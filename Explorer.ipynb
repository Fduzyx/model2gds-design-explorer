### Five-minute decision walkthrough

1. W1: compare minimum cycles and minimum STA latency, then choose ADP.
   The objective changes the preferred array; the cycle/STA order did not reverse.
2. W4: choose latency × area^α, set α = 0.95, and switch Core / Cell area.
   Inspect the different thresholds. Press **Set α to exact tie** to retain both
   candidates rather than force a winner. α describes preference, not a measurement.
3. Try the three area presets, then an exact numeric boundary. Area ≤ budget
   includes equality. Changing the area basis resets the numeric budget to its
   preset (Custom → Both fit); it never treats cell area as core area.
4. Select trace array 4×4 and Play. This is a current Python token trace checked
   against historical RTL, not a live RTL waveform. Open **Logic → physical
   evidence** for source-confirmed register locations, then the actual GDS image.
5. Expand evidence: only two fixed layouts and six fixed workloads, not six
   independent physical experiments. Pareto points are discrete alternatives.

STA latency = recorded RTL cycles × the fixed-layout internal R2R STA estimate.
It is not silicon Fmax, full IO closure or complete-system latency. Different
objectives, budgets or area definitions answer different design questions.
